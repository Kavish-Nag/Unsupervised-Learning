```python
from sklearn.datasets import load_wine

data = load_wine()
X = data.data
y = data.target
```

```python
import numpy as np
print(np.isnan(X).sum())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
import numpy as np

U, S, VT = np.linalg.svd(X_scaled)
```

```python
k = 2

U_k = U[:, :k]
S_k = np.diag(S[:k])
VT_k = VT[:k, :]

X_reduced = U_k @ S_k
```

```python
X_approx = U_k @ S_k @ VT_k
```

```python
energy = sum(S[:k]**2) / sum(S**2) #Energy retained
print(energy)
```

```python
error = np.linalg.norm(X_scaled - X_approx)
print(error)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X_reduced[:,0], X_reduced[:,1], c=y)
plt.title("SVD on Wine Dataset")
plt.show()
```
