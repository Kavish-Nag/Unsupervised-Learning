```python
from sklearn.metrics import DistanceMetric

X = [[1,2], [3,4], [5,6]]
Y = [[7,8], [9,10]]

dist = DistanceMetric.get_metric('euclidean')
result = dist.pairwise(X, Y)

print(result)
```

```python
dist = DistanceMetric.get_metric('manhattan')
print(dist.pairwise(X, Y))
```

```python
dist = DistanceMetric.get_metric('chebyshev')
print(dist.pairwise(X, Y))
```

```python
dist = DistanceMetric.get_metric('minkowski')
print(dist.pairwise(X, Y))
```

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from scipy.spatial.distance import euclidean, cityblock, minkowski

# Load dataset
data = load_iris()
X = data.data

# Preprocessing (Scaling)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Take 3 samples
A = X_scaled[0]
B = X_scaled[1]
C = X_scaled[2]

# Compute distances
print("\nDistances between Sample 0 and Sample 1:")
print("Euclidean Distance:", euclidean(A, B))
print("Manhattan Distance:", cityblock(A, B))
print("Minkowski Distance :", minkowski(A, B, p=3))

print("\nDistances between Sample 0 and Sample 2:")
print("Euclidean Distance:", euclidean(A, C))
print("Manhattan Distance:", cityblock(A, C))
print("Minkowski Distance :", minkowski(A, C, p=3))

print("\nDistances between Sample 1 and Sample 2:")
print("Euclidean Distance:", euclidean(B, C))
print("Manhattan Distance:", cityblock(B, C))
print("Minkowski Distance :", minkowski(B, C, p=3))
```
